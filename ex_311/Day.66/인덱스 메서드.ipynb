```python
import pandas as pd 
```

```python
df = pd.DataFrame({"이름": ["김민준", "이서연", "박지훈"], 
                   "점수": [85, 90, 75]}, 
                   index=["가", "나", "다"])
# index 매개값 생략하면 Integer 위치값(iloc)과 동일한 index값을 기본값으로 생성
print(df)
df # 기본값 display(df)
```

```python
# 인덱스 -> 열로 변경
df2 = df.reset_index()
df2
```

```python
# index -> 열
df.reset_index(inplace=True) # 기본값 False (원본변경 X)
df
```

```python
# 열 -> 인덱스 set_index
df.set_index("index", inplace=True)
df
```
