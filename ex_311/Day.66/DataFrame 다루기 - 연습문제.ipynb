```python
import pandas as pd
```

## 1. concat()을 사용하여 df1과 df2를 행 방향으로 결합하시오.

```python
df1 = pd.DataFrame({
    "이름": ["김민준", "이서연", "박지훈"],
    "점수": [85, 90, 75]
})

df2 = pd.DataFrame({
    "이름": ["최유진", "정현우", "강수빈"],
    "점수": [95, 80, 88]
})
```

```python
df3 = pd.concat([df1, df2], axis=0, ignore_index=True)
df3
```

## ​2. 두 DataFrame을 열 방향으로 결합하시오.

```python
df1 = pd.DataFrame({
    "이름": ["김민준", "이서연", "박지훈"],
    "점수": [85, 90, 75]
})

df2 = pd.DataFrame({
    "이름": ["최유진", "정현우", "강수빈"],
    "등급": ["B", "A", "C"]
})
```

```python
df4 = pd.concat([df1, df2], axis=1)
df4
```

## 3. 학생ID를 기준으로 두 DataFrame을 결합하시오.

```python
students = pd.DataFrame({
    "학생ID": [101, 102, 103, 104],
    "이름": ["김민준", "이서연", "박지훈", "최유진"]
})

scores = pd.DataFrame({
    "학생ID": [101, 102, 103, 104],
    "점수": [85, 90, 75, 95]
})
```

```python
result = pd.merge(
    students,
    scores,
    on="학생ID",
    how="inner"
)

result
```

## ​4. 고객ID를 기준으로 두 DataFrame을 merge()하시오.
(결과에 다음 컬럼이 포함되도록 하시오.)

```python
customer = pd.DataFrame({
    "고객ID": [1, 2, 3],
    "이름": ["김민준", "이서연", "박지훈"]
})

order = pd.DataFrame({
    "고객ID": [1, 2, 3],
    "상품": ["노트북", "마우스", "키보드"],
    "금액": [1200000, 30000, 80000]
})
```

```python
result2 = pd.merge(
    customer,
    order,
    on="고객ID",
    how="outer"
)

result2
```

## 5. 부서ID를 기준으로 두 DataFrame을 결합하여 다음과 같은 결과를 만드시오.

```python
employee = pd.DataFrame({
    "사번": [101, 102, 103, 104],
    "이름": ["김민준", "이서연", "박지훈", "최유진"],
    "부서ID": [10, 20, 10, 30]
})

department = pd.DataFrame({
    "부서ID": [10, 20, 30],
    "부서명": ["개발팀", "인사팀", "영업팀"]
})
```

```python
result3 = pd.merge(
    employee,
    department,
    on="부서ID",
    how="left",
)

result3
```
