```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
```

```python
# 60, 62, 65, 67, 68, 70, 72, 75, 78, 95, 200
arr = np.array([60, 62, 65, 67, 68, 70, 72, 75, 78, 95, 200])
print(arr)
print(type(arr))
```

```python
s = pd.Series(arr)
print(s)
print(type(s))
```

```python
# IQR = Q3 - Q1
# 중앙값 = Q2

Q1 = s.quantile(0.25)
Q3 = s.quantile(0.75)
median = s.median()
print("Q1:", Q1, "median:", median, "Q3:", Q3)

IQR = Q3 - Q1
print("IQR:", IQR)

# 상한/하한
upper = Q3 + (IQR*1.5)
lower = Q1 - (IQR*1.5)
print("upper:", upper)
print("lower:", lower)
# 60, 62, 65, 67, 68, 70, 72, 75, 78, 95, 200
```

```python
# 시각화 -> boxplot
fig = plt.figure()
axes1 = fig.add_subplot(1, 1, 1)
sns.boxplot(s)
plt.show()
```

```python
# 확실한 이상치(200) 제거 후, 다시 한번 진행 -> 두번째도 이상치가 95이면... -> 이상치가 발션되지 않을 때까지...
```

```python
# Z-score : 평균에서 얼마나 멀리 있는가?

mean = s.mean()
print(s)
print("mean:", mean)

# 분산 / 표준편차
var = round(arr.var(), 1)
std = round(arr.std(), 1)
print("var:", var, "std:", std)

# 평균 82.9
# 60, 62, 65, 67, 68, 70, 72, 75, 78, 95, 200 
#                               -0.4   0   1     -> Z-score
#
# (각각의 값 - 평균) + ... / std
print(s-82.9)
print((s - 82.9).sum() / std)

Z_score = (s-mean) / std
print(Z_score)
print(Z_score.abs() > 3)

# 시각화 -> histplot
fig = plt.figure()
axes1 = fig.add_subplot(1, 1, 1)
sns.histplot(s, bins=5)
plt.show()
```
