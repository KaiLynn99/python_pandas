## sklearn의 KBinsDiscretizer클래스를 이용한 이산화 실습

```python
import pandas as pd
from sklearn.preprocessing import KBinsDiscretizer

# 데이터
df = pd.DataFrame({
    "나이": [15, 23, 35, 47, 62]
})

display(df)

# 이산화
kb = KBinsDiscretizer(
    n_bins=3,              # 3개 구간
    encode="ordinal",      
    strategy="uniform"    
)


df["나이_구간"] = kb.fit_transform(df[["나이"]])

display(df)
```
